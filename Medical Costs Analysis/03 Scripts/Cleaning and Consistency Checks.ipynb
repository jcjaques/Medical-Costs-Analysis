```python
# checking for mixed data types
for col in df_Insur.columns.tolist():
    weird = (df_Insur[[col]].map(type) != df_Insur[[col]].iloc[0].apply(type)).any(axis = 1)
    if len (df_Insur[weird]) > 0:
        print(col)
```
